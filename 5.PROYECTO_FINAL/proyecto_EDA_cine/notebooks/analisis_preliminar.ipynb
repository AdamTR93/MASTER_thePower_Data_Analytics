Tratamiento de presupuesto y recaudación

El presupuesto `budget` está a cero en 9.108 de las 17.748 películas (un 51%). Se confirma que este valor representa un dato no informado y no un presupuesto real: películas como *Your Name.*, *Suzume* o *Monster Hunt* aparecen con presupuesto 0 pese a haber recaudado más de 300 millones de dólares. En la recaudación `revenue`, el cero tiene además un matiz: entre las películas sin recaudación aparecen estrenos directos en plataformas como *To All the Boys I've Loved Before* o *El Camino: A Breaking Bad Movie*, para las que la taquilla simplemente no aplica. En ambos casos, mantener el cero distorsionaría las medias de recaudación y rentabilidad, por lo que se convierte a nulo.

Se han detectado también valores positivos pero inverosímiles, por debajo de 1.000 dólares, en unas 37 películas en el caso del presupuesto y 35 en el de la recaudación. La mayoría parecen cifras introducidas en millones (por ejemplo, *Tom and Jerry: The Movie* con un presupuesto de 35 dólares o *City Slickers II* con 40), aunque existen excepciones reales como *Who Killed Captain Alex?*, película ugandesa rodada con apenas unos cientos de dólares. Al no poder determinar con certeza la escala correcta de cada caso, se ha optado por un criterio conservador y homogéneo: todos los valores inferiores a 1.000 dólares se convierten a nulo. Una posible mejora futura sería verificar y corregir manualmente estos casos con fuentes externas.

Por último, se crea la variable `has_financial_data`, que indica si una película cuenta con presupuesto y recaudación válidos.

Tras la limpieza, 7.049 películas (un 39,7% del dataset) disponen de información financiera completa. Esta variable permite conservar todas las películas para el análisis de valoraciones, géneros y duraciones, y filtrar solo las que tienen datos financieros en el análisis de rentabilidad. Dado que más de la mitad de los valores financieros están ausentes, se descarta imputarlos, ya que introduciría un sesgo importante en una de las variables clave del estudio.

## 5. Creación de nuevas variables

### `release_year`, `release_month` y `decade`

```python
df_clean['release_year'] = df_clean['release_date'].dt.year
df_clean['release_month'] = df_clean['release_date'].dt.month
df_clean['decade'] = (df_clean['release_year']//10)*10

print(df_clean[['release_date', 'release_year', 'release_month', 'decade']].head(10))
print('-----------------------------------------------------')
df_clean['decade'].value_counts().sort_index(ascending=False)
```

A partir de la fecha de estreno `release_date` se crean tres nuevas variables: el año de estreno `release_year`, el mes de estreno `release_month` y la década `decade`, que permitirán analizar la evolución temporal y la estacionalidad de los estrenos.

La distribución por décadas muestra un fuerte sesgo hacia el cine reciente: la década de 2010 concentra 6.765 películas, el 38% del dataset, más que todas las décadas anteriores a 1990 juntas. Esto se debe tanto al aumento de la producción cinematográfica como al perfil de los usuarios de TMDB, que valoran sobre todo películas recientes, lo que hace que las antiguas alcancen con menos frecuencia el mínimo de 100 votos. Además, la década de 2020 está incompleta, ya que solo incluye datos hasta septiembre de 2023, y las primeras décadas del siglo XX cuentan con muy pocas películas, por lo que las comparaciones entre décadas deberán tener en cuenta estas limitaciones.
