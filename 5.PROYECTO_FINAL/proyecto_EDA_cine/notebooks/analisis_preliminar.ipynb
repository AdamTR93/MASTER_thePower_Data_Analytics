## 3. Inspección de columnas

Habiendo hecho un primer vistazo al dataset, vamos a quedarnos con las columnas que nos aportarán información necesaria para el EDA y vamos a prescindir de las que no necesitamos.

```python
columnas_drop = ['backdrop_path', 'homepage', 'imdb_id', 'original_title', 'overview', 'tagline', 'keywords', 'spoken_languages']
df_clean = df_cine.drop(columns = columnas_drop)
df_clean.shape
```

`status`

```python
df_clean['status'].value_counts()
```

nos vamos a quedar solo con las peliculas`Released' ya que las películas en producción, planificadas, rumoreadas o canceladas todavía no se han estrenado, así que no pueden tener recaudación, votos ni nota reales.

```python
# seleccionamos solo las peliculas released y despues borramos esa columna ya que no la necesitamos
df_clean = df_clean[df_clean['status'] == 'Released'].copy()
df_clean = df_clean.drop(columns= 'status')
```

```python
df_clean.shape
```

`adult`

```python
df_clean['adult'].value_counts()
```

```python
adult_df = df_clean[df_clean['adult']]

pct_budget_0 = (adult_df['budget'] == 0).mean() * 100
pct_revenue_0 = (adult_df['revenue'] == 0).mean() * 100
pct_ambos_0 = ((adult_df['budget'] == 0) & (adult_df['revenue'] == 0)).mean() * 100

print(f'Películas adultas con budget 0: {pct_budget_0:.2f}%')
print(f'Películas adultas con revenue 0: {pct_revenue_0:.2f}%')
print(f'Películas adultas con ambos a 0: {pct_ambos_0:.2f}%')
```

```python
df_clean = df_clean[df_clean['adult'] != True].copy()
df_clean = df_clean.drop(columns = 'adult')
df_clean.shape
```

Se han filtrado las pelicuals con contenido para adultos y han salido 150908, casi un 10% del dataset. Al profundizar un poco en la variable,
el análisis de sus datos financieros confirma que el 99,15% tiene el presupuesto `budget` a 0, el 99,74% tiene la recaudación `revenue` a 0, y el 99,13% carece de ambos datos a la vez.

Dado que el objetivo del análisis es entender qué hace rentable y bien valorada a una película comercial, se ha decidido eliminar estos registros, ya que no aportan información financiera y distorsionarían el estudio de géneros, valoraciones y duraciones. Tras este filtro, el dataset pasa de 1.477.088 a 1.326.180 películas, y la columna adult se elimina al contener ya un único valor.

`release_date`

```python
df_no_date = df_clean[df_clean['release_date'].isna()]

pct_no_date = df_clean['release_date'].isna().mean()*100
no_date_budget_0 = (df_no_date['budget'] == 0).mean() * 100
no_date_revenue_0 = (df_no_date['revenue'] == 0).mean() * 100
no_date_ambos_0 = ((df_no_date['budget'] == 0) & (df_no_date['revenue'] == 0)).mean() * 100
no_date_vote_0 = (df_no_date['vote_count'] == 0).mean()*100
no_date_best_pop = df_no_date[['popularity','title','genres']].sort_values(by='popularity',ascending=False)

print(f'Películas sin fecha de lanzamiento: {pct_no_date:.2f}%')
print(f'Películas sin fecha de lanzamiento con budget 0: {no_date_budget_0:.2f}%')
print(f'Películas sin fecha de lanzamiento con revenue 0: {no_date_revenue_0:.2f}%')
print(f'Películas sin fecha de lanzamiento con ambos a 0: {no_date_ambos_0:.2f}%')
print(f'Películas sin fecha de lanzamiento con 0 votos: {no_date_vote_0:.2f}%')
no_date_best_pop.head(20)
```

```python
df_clean = df_clean.dropna(subset = ['release_date'])
df_clean.shape
```

La fecha de estreno release_date no está informada en el 24,11% de las películas restantes. Además, al ordenarlas por popularidad `popularity`, los títulos más destacados no corresponden a películas comerciales sino a retransmisiones de eventos, conciertos, documentales y recopilatorios, con valores de popularidad muy bajos respecto al máximo del dataset.

Por todo ello, y dado que sin fecha de estreno estas películas no pueden situarse en el tiempo (un eje clave del análisis), se ha decidido eliminarlas del dataset.

```python
fecha_hoy = pd.Timestamp.today()

fechas_bajas = (df_clean['release_date'] < '1895-01-01').sum()
fechas_altas = (df_clean['release_date'] > fecha_hoy).sum()

print(f'Películas anteriores a 1895: {fechas_bajas}')
print(f'Películas con fecha futura: {fechas_altas}')
```

Tras convertir la fecha de estreno `release_date` a formato fecha, el rango de valores va desde el año 1800 hasta 2061, lo que revela registros incoherentes en ambos extremos. Por un lado, 229 películas tienen fecha anterior a 1895, año de la primera proyección pública de los hermanos Lumière. Entre ellas aparecen errores evidentes con la fecha de relleno 1 de enero de 1800, pero también obras históricas auténticas de precine, como las secuencias fotográficas de Eadweard Muybridge de 1878, que no corresponden a películas en sentido comercial. Por otro lado, 104 películas marcadas como estrenadas tienen una fecha posterior a la actual, como proyectos todavía sin título o temporadas de ópera futuras, lo que indica errores en la introducción de datos.

Se ha decidido conservar únicamente las películas estrenadas entre 1895 y la fecha actual, eliminando 333 registros (un 0,03% del total). Durante esta revisión se ha detectado además la existencia de títulos duplicados con distinto identificador id, que se tratarán en un paso posterior.
