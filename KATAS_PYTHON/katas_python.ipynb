## 1. *Escribe una función que reciba una cadena de texto como parámetro y devuelva un diccionario con las frecuencias de cada letra en la cadena. Los espacios no deben ser considerados.*

```python
def contador_letras(frase):
    """
    funcion para contar letras

    Args:
        frase (str): frase que se introduce para contar las letras

    Returns:
        dict: diccionario con la letra y el número de veces que aparece en la frase
    """    
    dict_frecuencia = {}

    for letra in frase.lower():
        if letra == ' ':
            continue
        elif letra in dict_frecuencia:
            dict_frecuencia[letra]+=1
        else:
            dict_frecuencia[letra]= 1
    return dict_frecuencia

# usamos la función creada
texto = input(f'Escribe una frase y contaremos todas sus letras:')
resultado = contador_letras(texto)
print(f'Esta es tu frase: "{texto}"')
print(f'Contador de letras: {resultado}')
```

## 2. *Dada una lista de números, obtén una nueva lista con el doble de cada valor. Usa la función map()*

```python
def cuadrado(num):
    """
    funcion para elevar al cuadrado un número

    Args:
        num (float o int): número introducido

    Returns:
        float o int: nos devueve el cuadrado del número proporcionado
    """    
    return num**2

# elevamos al cuadrado una lista con map()
lista_num = [1,2,3,4,5,6,7,8]
print(f'Estos son los números proporcionados: {lista_num}')
resultado = list(map(cuadrado,lista_num)) # tenemos que convertirlo en una lista para que nos muestre el resultado
print(f'Estos son los números proporcionados al cuadrado: {resultado}')
```

## 4. *Genera una función que calcule la diferencia entre los valores de dos listas. Usa la función map()*

```python
def diferencia_listas(lista1,lista2):
    resultado = []
    for l1,l2 in zip(lista1,lista2):
        resultado.append(l1-l2)
    return resultado
        

lista_num1 = [8,7,30,17]
lista_num2 = [1,3,5,7]

resultado = diferencia_listas(lista_num1,lista_num2)

print(resultado)
```

## 5. *Ecribe una función que tome una lista de números como parámetro y un valor opcional nota_aprobado, que por defecto es 5. La función debe calcular la media de los números en la lista y determinar si la media es mayor o igual que nota aprobado. Si es así, el estado será "aprobado", de lo contrario, será "suspenso". La función debe devolver una tupla que contenga la media y el estado.*

```python
def notas(lista_num, nota_aprobado = 5): # ponemos nota_aprobado para que tome 5 por defecto siempre que usemos la funcion
    """
    función que calcula la media de las notas y te dice si has aprobado o suspendido

    Args:
        lista_num (list): Lista de números (notas) a promediar
        nota_aprobado (int, optional): nota que marca el aprobado. Defaults to 5.
    Returns:
        tuple: muestra la nota media y si has supendido o aprobado.
    """
    media = (sum(lista_num)/len(lista_num)) 
    if media >= nota_aprobado:
        estado = '¡aprobado!'
    else:
        estado = 'supendido'
    return (media,estado)

valores = [6,7,8,9,4]
nota_media, resultado = notas(valores)
print(f'Tu nota media es de {nota_media} has {resultado}')
```

## 6. *Escribe una función que calcule el factorial de un número de manera recursiva*

```python
def factorial(num):
    """
    funcion para calcular el factorial de un número int

    Args:
        num (int): número proporcionado

    Returns:
        número int: el factorial del número proporcionado
    """    
    if num <= 1:
        return 1
    else:
        return num*factorial(num-1)

factorial(6)
```

## 9. *Escribe una función que tome una lista de nombres de mascotas como parámetro y devuelva una nueva lista excluyendo ciertas mascotas prohibidas en España. La lista de mascotas a excluir es ["Mapache", "Tigre", "Serpiente Pitón", "Cocodrilo", "Oso"].Usa la función filter()*

```python

def animales_excluidos(animal):
    """Comprueba si un animal es legal en España, ignorando espacios y mayúsculas.

    Args:
        animal (str): nombre de animal para saber si está prohibido o no.

    Returns:
        bool: True si el animal está permitido, False si está prohibido.
    """    
    animales_prohibidos = ["mapache", "tigre", "serpiente pitón", "cocodrilo", "oso"]
    return animal.strip().lower() not in animales_prohibidos

def animales_prohibidos(lista_animales):
    """Filtra una lista completa de animales usando la función filter().

    Args:
        lista_animales (list): lista con nombres de animales proporcionada por el usuario.

    Returns:
        list: nueva lista sin los animales prohibidos
    """    
    # filter() aplica la regla y list() convierte el resultado de nuevo en lista
    lista_filter = list(filter(animales_excluidos,lista_animales))
    return lista_filter

input_user = input('Pon tus animales favoritos separados por comas:').lower()
lista_animales = input_user.split(',')

resultado = animales_prohibidos(lista_animales)

print(f'Estos son los animales permitidos en España de entre todos los que has puesto:{resultado}')
```

 ## 10. *Escribe una función que reciba una lista de números y calcule su promedio. Si la lista está vacía, lanza una excepción personalizada y maneja el error adecuadamente.*


```python
def promedio_lista(lista_num):
    """Calculadora de promedio de una lsita de números

    Args:
        lista_num (list): lista de números.

    Raises:
        ValueError: Error que salta cuando la lista está vacía.

    Returns:
        float: resultado de la media de la lista de números.
    """    
    if len(lista_num) == 0:
        raise ValueError('¡Error! Tienes que intorducir al menos un número.')
    else:
        suma_total = 0
        for i in lista_num:
            suma_total+=i
        # redondeamos a 2 decimales    
        return round(suma_total/len(lista_num),2)

try:
    input_user = input('Introduce números separados por comas para saber la media:')
    # Evita que python de un error si el usuario no escribe nada.
    if input_user.strip() == "":
        lista_numeros = []
    else:
        lista_numeros = list(map(int,input_user.split(',')))

    resultado = promedio_lista(lista_numeros)
    print(f' La media de los números que has introducido es de : {resultado}')

except ValueError as lista_vacia:
    # Muestra el error personalizado
    print(lista_vacia)

```

## 17. *Crea una función que tome una lista de dígitos y devuelva el número correspondiente. Por ejemplo, [5,7,2]corresponde al número quinientos setenta y dos [572]. Usa la función reduce()*

```python
# Primero importamos la función reduce
from functools import reduce

def conversor_num(multiplicado, suma_siguiente):
    """desplaza el valor acumulado una posición a la izquierda (multiplicando por 10) y le suma el siguiente dígito de la lista

    Args:
        multiplicado (int): el número entero acumulado hasta el momento
        suma_siguiente (int): el siguiente dígito a sumar

    Returns:
        int: el nuevo número entero combinado
    """
    return (multiplicado * 10) + suma_siguiente

numeros = [5,7,2]
# reduce() recorre la lista aplicando la función de forma acumulativa de izquierda a derecha.
resultado = reduce(conversor_num, numeros)
print(resultado)
```

## 28. *Crea una función que busque y devuelva el primer elemento duplicado en una lista dada*

```python
def repetidos(lista_repes):
    """busca y devuelve el primer elemento que se encuentre duplicado en una lista

    Args:
        lista_repes (list): la lista de elementos (números o textos) donde buscar

    Returns:
        str: un mensaje indicando cuál es el primer elemento duplicado encontrado
    """    
    repes = set()
    for e in lista_repes:
        if e in repes:
            return f'¡Duplicado encontrado! Es "{e}"'
        else:
            repes.add(e)

elementos = ['perro','gato',4,9,'gato']

resultado = repetidos(elementos)

print(resultado)
```

## 29. *Crea una función que convierta una variable en una cadena de texto y enmascare todos los caracteres con el carácter '#', excepto los últimos cuatro.*

```python
def oculto(elemento):
    """Enmascara una variable ocultando sus caracteres excepto los últimos cuatro

    Args:
        elemento (any): La variable, número o texto que se desea enmascarar

    Returns:
        str: Una cadena de texto con los caracteres iniciales ocultos con '#' y los últimos cuatro visibles
    """    
    elemento_txt = str(elemento)
    num_oculto = len(elemento_txt) - 4
    texto_oculto = '#' * num_oculto
    texto_visible = elemento_txt[-4:]

    return texto_oculto + texto_visible

variable = 123456789

resultado = oculto(variable)

print(resultado)
```

```python
def anagrama(palabra1,palabra2):
    """Determina si dos cadenas de texto son anagramas (palabras diferentes escritas con las mismas letras)

    Args:
        palabra1 (str): La primera palabra o frase a comparar
        palabra2 (str): La segunda palabra o frase a comparar

    Returns:
        bool: True si ambas palabras contienen exactamente las mismas letras o False en caso contrario
    """ 
    palabra1 = palabra1.lower().replace(' ','')
    palabra2 = palabra2.lower().replace(' ','')

    # sorted() ordena las letras alfabéticamente; si las listas resultantes son idénticas, significa que comparten exactamente las mismas letras
    return sorted(palabra1) == sorted(palabra2)

print('Escribe dos palabras para descubrir si son anagramas')
p1 = input('Escribe una palabra:')
p2 = input('Escribe una segunda palabra:')

if anagrama(p1, p2):
    print(f'¡Sí! "{p1}" y "{p2}" son anagramas.')
else:
    print(f'No, "{p1}" y "{p2}" no son anagramas.')
```

## 34. *Crea la clase Arbol , define un árbol genérico con un tronco y ramas como atributos. Los métodos disponibles son:crecer_tronco , nueva_rama , crecer_ramas , quitar_rama e info_arbol . El objetivo es implementar estos métodos para manipular la estructura del árbol.*

```python
class Arbol:
    """representa un árbol genérico
    """ 
    # MÉTODO CONSTRUCTOR   
    def __init__(self):
        """iniciamos nuestro arbol con tronco de longitud 1 y sin ramas
        """         
        self.tronco = 1
        self.ramas = []
    
    # PRIMER MÉTODO: CRECER TRONCO
    def crecer_tronco(self):
        """aumenta 1 unidad el tamño del tronco
        """              
        self.tronco+=1
        print(f'¡El tronco ha crecido! Nueva longitud del tronco: {self.tronco}.')
    
    # SEGUNDO MÉTODO: AGREGAR NUEVA RAMA
    def nueva_rama(self):
        """añade una rama de longitud 1 a la lista de ramas
        """        
        self.ramas.append(1)
        print(f'¡Ha nacido una nueva rama! El arbol tiene: {len(self.ramas)} ramas ')
    
    # TERCER MÉTODO: ALARGAR RAMAS
    def crecer_ramas(self):
        """suma 1 a todas las ramas que haya en la lista en ese momento
        """        
        self.ramas = [e +1 for e in self.ramas]

        print(f'¡Las ramas han crecido! Nueva longitud: {self.ramas}.')
    
    # CUARTO MÉTODO: ELIMINAR RAMA 

    def quitar_rama(self,posicion):
        """elimina una rama de la lista según la posición seleccionada

        Args:
            posicion (int): índide de la rama que se queire eliminar
        """        
        self.ramas.pop(posicion)
        print(f'¡Una rama cayó! Nuestro árbol tiene {len(self.ramas)} ramas')
    
    # QUINTO MÉTODO: INFO DEL ÁRBOL

    def info_arbol(self):
        print(f'---------- INFORMACIÓN DEL ÁRBOL ----------------')
        print(f'- Longitud del tronco: {self.tronco} unidades.')
        print(f'- Número total de ramas: {len(self.ramas)} ramas.')
        print(f'- Longitud de cada rama: {self.ramas}')


# ---------------------- USO DE LA CLASE ÁRBOL -----------------------------
# 1. Crear un árbol.   

mi_arbol = Arbol()
print(f'Este es nuestro arbol: {mi_arbol.__dict__}')

# 2. Hacer crecer el tronco del árbol una unidad.

mi_arbol.crecer_tronco()

# 3. Añadir una nueva rama al árbol.

mi_arbol.nueva_rama()
print(f'Arbol con ramas: {mi_arbol.ramas}')

# 4. Hacer crecer todas las ramas del árbol una unidad.

mi_arbol.crecer_ramas()
print(f'Longitud de cada rama: {mi_arbol.ramas}')

# 5. Añadir dos nuevas ramas al árbol.

mi_arbol.nueva_rama()
mi_arbol.nueva_rama()
print(f'Arbol con ramas: {mi_arbol.ramas}')

# 6. Retirar la rama situada en la posición 2.
print(f'Nuestro árbol tiene {len(mi_arbol.ramas)} ramas')
mi_arbol.quitar_rama(2)

# 7. Obtener información sobre el árbol.

mi_arbol.info_arbol()

```

## 35. *Crea la clase UsuarioBanco ,representa a un usuario de un banco con su nombre, saldo y si tiene o no cuenta corriente. Proporciona métodos para realizar operaciones como retirar dinero, transferir dinero desde otro usuario y agregar dinero al saldo.*

```python
class UsuarioBanco:
    """representa el usuario de un banco
    """    
    # MÉTODO CONSTRUCTOR

    def __init__(self,nombre,saldo,cuenta_corriente):
        """constructor de la clase UsuarioBanco

        Args:
            nombre (str): nombre del usuario
            saldo (float): saldo inicial del usuario
            cc (bool): True si tiene cuenta corriente o False si no la tiene.
        """       

        self.nombre = nombre
        self.saldo = saldo
        self.cuenta_corriente = cuenta_corriente 
    
    # PRIMER MÉTODO: RETIRAR DINERO
    def retirar_dinero(self,cantidad):
        """retira una cantidad de dinero del saldo del usuario

        Args:
            cantidad (float): la cantidad de dinero que se desea retirar

        Raises:
            ValueError: Si el usuario no tiene suficiente saldo disponible
        """        
        if cantidad > self.saldo:
            raise ValueError ('¡Error! no tienes dinero suficiente en la cuenta')
        else:
            self.saldo = self.saldo - cantidad
            return(f'¡Operación existosa! Se han retirado {cantidad} de la cuenta. El saldo de {self.nombre} es de {self.saldo}')

    # SEGUNDO MÉTODO: TRANSFERIR DINERO
    def transferir_dinero(self,otro_user,cantidad):
        """realiza una transferencia desde un usuario hacia el usuario actual

        Args:
            otro_usuario (UsuarioBanco): el objeto del usuario que envía el dinero
            cantidad (float): la cantidad de dinero a transferir

        Raises:
            ValueError: si alguno de los usuarios no tiene cuenta corriente o 
                        si el usuario que envía dinero, no tiene saldo suficiente
        """        
        if not self.cuenta_corriente or not otro_user.cuenta_corriente:
            raise ValueError('¡Error! Ambos usuarios deben tener una cuenta corriente activa.')
        otro_user.retirar_dinero(cantidad)
        self.agregar_dinero(cantidad)
    
    # TERCER MÉTODO: AGREGAR DINERO
    def agregar_dinero(self,cantidad):
        """añade una cantidad de dinero al saldo actual del usuario

        Args:
            cantidad (float): la cantidad de dinero que se va a ingresar al saldo actual
        """        
        self.saldo = self.saldo + cantidad
        return(f'¡Operación existosa! Añadidos {cantidad} a la cuenta. El saldo de {self.nombre} es de {self.saldo}')
    




# 1. Crear dos usuarios: "Alicia" con saldo inicial de 100 y "Bob" con saldo inicial de 50, ambos con cuenta corriente.

alicia = UsuarioBanco('Alicia',100,True)
bob = UsuarioBanco('Bob',50,True)

# 2. Agregar 20 unidades de saldo de "Bob".

bob.agregar_dinero(20)

# 3. Hacer una transferencia de 80 unidades desde "Bob" a "Alicia".

try:
    alicia.transferir_dinero(bob,80)
except ValueError as e:
    print(e)


# 4. Retirar 50 unidades de saldo a "Alicia".

alicia.retirar_dinero(50)
```

```python
texto = input('Escribe un texto:').lower()

def contar_palabras(texto):
    """cuenta el número total de palabras en un texto

    Args:
        texto (str): el string original
    Returns:
        int: número de palabras.
    """
    palabras_texto = texto.split()
    return len(palabras_texto)
    
def reemplazar_palabras(texto,palabra_vieja,palabra_nueva):
    """Busca una palabra y la cambia por otra

    Args:
        texto (str): El string original
        palabra_vieja (str): la palabra que queremos quitar
        palabra_nueva (str): la palabra que queremos poner
    Returns:
        str: El texto modificado
    """    
    cambio_palabra = texto.replace(palabra_vieja,palabra_nueva)
    return cambio_palabra

def eliminar_palabra(texto,palabra_out):
    """Elimina por completo una palabra del texto

    Args:
        texto (str): El string original
        palabra_out (str): La palabra a borrar
    Returns:
        str: El texto sin la palabra
    """    
    palabra_eliminada = texto.replace(palabra_out,"")
    return palabra_eliminada

def procesar_texto(texto,funcion,*args):
    """Procesa un texto aplicando la función que se le pase por argumento.

    Args:
        texto (str): El texto a procesar.
        funcion (function): La función específica que queremos ejecutar.
        *args: Argumentos extra que pueda necesitar la función (palabras a cambiar/borrar).
    Returns:
        El resultado de ejecutar la función elegida.
    """    
    return funcion(texto,*args)

resultado_contar = procesar_texto(texto,contar_palabras)
resultado_reemplazar = procesar_texto(texto,reemplazar_palabras,'adam','pedro')
resultado_eliminar = procesar_texto(texto,eliminar_palabra,'adam')
print(f'Frase: {texto}')
print(f'1. Cantidad total de palabras: {resultado_contar}')
print(f'2. Texto tras reemplazar "adam" por "pedro": "{resultado_reemplazar}"')
print(f'3. Texto tras eliminar la palabra "adam": "{resultado_eliminar}"')
```

```python
def area_rectangulo(datos):
    """calcula el área de un rectangulo

    Args:
        datos (tuple): tupla con la base y la altura

    Returns:
        int/float: el área del rectangulo calculada
    """    
    base,altura = datos
    return base * altura

def area_triangulo(datos):
    """calcula el área de un triangulo

    Args:
        datos (tuple): tupla con la base y la altura
    Returns:
        int/float: el área del triangulo calculada
    """    
    base,altura = datos
    return (base*altura)/2

def area_circulo(datos):
    """calcula el área de un circulo

    Args:
        datos (tuple): tupla con un único elemento, el radio del circulo.

    Returns:
        float: el área del circulo calculada
    """    
    radio, = datos
    return 3.1416 * radio**2

def calculo_figura(figura,datos):
    """recibe el nombre de la figura y nos lleva a la función correcta para calcular su área

    Args:
        figura (str): el nombre de la figura ('rectangulo', 'triangulo' o 'circulo')
        datos (tuple): una tupla con las medidas necesarias para el cálculo

    Returns:
        float o str: el área calculada de la figura o un mensaje de error si no es válida
    """   
    figura = figura.lower()

    if figura == 'rectangulo':
        return area_rectangulo(datos)
    
    elif figura == 'triangulo':
        return area_triangulo(datos)
    
    elif figura == 'circulo':
        return area_circulo(datos)
    
    else:
        return f'¡Error! No se pude calcular el area de esa figura.'
   
# COMPROBAMOS ÁREA RECTANGULO base 20 y altura 10

print(f" Área del rectangulo: {calculo_figura('rectangulo',(20,10))}")

# COMPROBAMOS ÁREA TRIANGULO base 17 y altura 33
print(f" Área del triangulo: {round(calculo_figura('triangulo',(17,33)),2)}")

# COMPROBAMOS ÁREA circulo radio 10
print(f" Área del circulo: {round(calculo_figura('circulo',(10,)),2)}")
```
